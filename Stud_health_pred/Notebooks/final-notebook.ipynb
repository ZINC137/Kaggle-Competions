```python
import pandas as pd
from collections import Counter

PATH = "/kaggle/input/datasets/smrutiranjanghosh/submissions/"

sub2 = pd.read_csv(PATH + "submission-2.csv")
sub3 = pd.read_csv(PATH + "submission-3.csv")
sub4 = pd.read_csv(PATH + "submission-4.csv")
sub5 = pd.read_csv(PATH + "submission-5.csv")

ensemble = pd.DataFrame()
ensemble["id"] = sub2["id"]

ensemble["sub2"] = sub2["health_condition"]
ensemble["sub3"] = sub3["health_condition"]
ensemble["sub4"] = sub4["health_condition"]
ensemble["sub5"] = sub5["health_condition"]


def majority_vote(row):
    votes = Counter(row)
    return votes.most_common(1)[0][0]


ensemble["health_condition"] = ensemble[
    ["sub2", "sub3", "sub4", "sub5"]
].apply(majority_vote, axis=1)

submission = ensemble[["id", "health_condition"]]

submission.to_csv("submission.csv", index=False)

print("Ensemble submission created!")
print(submission.head())
```
